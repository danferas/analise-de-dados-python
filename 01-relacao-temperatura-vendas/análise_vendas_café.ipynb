# Exercício 2 – Análise de Vendas de Café

Neste exercício, trabalhamos com uma base de dados de vendas de café e temperatura ao longo de uma semana.

O objetivo foi organizar os dados, observar algumas informações estatísticas e criar um gráfico para comparar as vendas com a temperatura.

```python
# Criar a base de dados simples
# Importar bibliotecas
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

dados = {'Dia': ['Dom', 'Seg', 'Ter', 'Qua', 'Qui', 'Sex', 'Sáb'],
         'Cafés_Vendidos': [50, 55, 50,  60, 65, 180, 70],
         'Temperatura': [22, 20, 21, 19, 18, 15, 17],
        }
df = pd.DataFrame(dados)
display(df)
display(df.describe().round(2))

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Temperatura',       # Eixo X: Temperatura (quanto mais à direita, mais calor)
    y='Cafés_Vendidos',    # Eixo Y: Cafés vendidos (quanto mais para cima, mais vendas)
    hue='Dia',   # Cor dos pontos: Muda de acordo com o dia da semana
    s=120,                 # Tamanho dos pontos para ficarem fáceis de ver
    palette='deep'         # Paleta de cores agradável
    )

# 3. Adicionando títulos e melhorando a estética
plt.title('Relação entre Temperatura, Vendas de Café e Dias da Semana', fontsize=14, fontweight='bold')
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Cafés Vendidos (unidades)', fontsize=12)

# Jogando a legenda para fora do gráfico para não tampar os pontos
plt.legend(title='Dia', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adicionando linhas de grade quadriculada para facilitar a leitura
plt.grid(True, linestyle='--', alpha=0.5)

# 4. Exibindo o gráfico limpo
plt.tight_layout()
plt.show()
```
