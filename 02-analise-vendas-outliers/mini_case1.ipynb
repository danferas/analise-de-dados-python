# Mini-Case 1: O Mistério do Bônus de Vendas

### Análise de Vendas: Média, Mediana e Outliers

**Objetivo:** Analisar os valores de vendas de uma equipe e entender como uma venda muito alta pode influenciar a média e a definição de metas.

## 1. Cenário do Problema

Uma loja possui 5 vendedores. Durante o mês de outubro, eles realizaram as seguintes vendas:

- Vendedor 1: R$ 10.000

- Vendedor 2: R$ 11.000

- Vendedor 3: R$ 12.000

- Vendedor 4: R$ 11.000

- Vendedor 5: R$ 56.000

Um dos vendedores fechou um contrato muito alto, fazendo com que a média de vendas chegasse a R$ 20.000.

O gerente pretende utilizar esse valor como meta para todos os vendedores no mês seguinte.

**Pergunta:** Essa é uma meta justa? Qual medida estatística representa melhor as vendas da equipe?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

dados = {'Vendas':[10000, 11000, 12000, 11000, 56000]}
df = pd.DataFrame(dados)
display(df)

media = df['Vendas'].mean()
mediana = df['Vendas'].median()
amplitude = df['Vendas'].max() - df['Vendas'].min()

print(f"Média: {media}")
print(f"Mediana: {mediana}")
print(f"Amplitude: {amplitude}")

plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df['Vendas'],
    color='lightcoral',
    width=0.4
)

plt.title(
    'Análise de Vendas da Equipe: O Impacto do Outlier',
    fontsize=12,
    fontweight='bold'
)

plt.ylabel('Valor de Vendas (R$)', fontsize=10)

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## Conclusão

A venda de R$ 56.000 fez a média subir bastante.

A mediana de R$ 11.000 mostra melhor quanto os vendedores costumam vender.

Por isso, não usaria a média de R$ 20.000 como meta para todos, pois ela não representa as vendas da maioria.
